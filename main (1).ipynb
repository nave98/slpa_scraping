```python
import requests
from bs4 import BeautifulSoup
import lxml
```

```python
url=("https://www.slpa.lk/port-colombo-list/news")
response=requests.get(url)
soup=BeautifulSoup(response.content, "lxml")
print(soup.prettify())
```

```python
#Latest News

news=soup.find_all("div",class_="content")
print(news)
```

```python
headers=[]

for i in news:
   headers.append(i.select_one("h3").get_text(strip=True))
print(headers)
```

```python
content=[]

for item in news:
    content.append(item.select_one("p").get_text(strip=True))
print(content) 
```

```python
read_more_links=[]

for i in news:
    read_more_links.append(i.select_one("a").get("href"))
print(read_more_links)
```

```python
news_dict = {
    "Headers":headers,
    "Content": content,
    "Read More": read_more_links
}

print(news_dict)
```

```python
import pandas as pd

news_df = pd.DataFrame(news_dict)
news_df
```

```python
news_df.to_csv("slpa_news.csv",index=False)
```

Foreign Procuretment Details


```python
import requests
from bs4 import BeautifulSoup
```

```python
url="https://www.slpa.lk/port-colombo/foreign-purchases"
response= requests.get(url)
soup=BeautifulSoup(response.content,"lxml")
print(soup.prettify())
```

FOREIGN-PURCHASES->Electrical equipment

```python
table = soup.find_all("tbody")[0]
print(table)
```

```python
titles=table.find_all("strong")
titles
```

```python
table_titles =[title.text for title in titles ]
table_titles
```

```python
import pandas as pd

df=pd.DataFrame(columns=table_titles)
df
```

```python
column_data = table.find_all("tr")
column_data
```

```python
for row in column_data:
    row_data=row.find_all("td")
    individual_row_data = [data.text.strip() for  data in row_data]
    print(individual_row_data)

    length= len(df)
    # identify the specific columns and rows
    df.loc[length]=individual_row_data
    
```

```python
df[1:]
```

```python
df[1:].to_csv("Foreing_Electrical_items_purchase.csv",index=False)
```
